## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d : {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

```python
print((df['impressions_90d'] > 0).value_counts())
print("If there's only one unique value (True or False), the standard deviation is zero, leading to a NaN correlation.")
```

```python
print((df['impressions_90d'] > 0).value_counts())
print("If there's only one unique value (True or False), the standard deviation is zero, leading to a NaN correlation.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

```python
worst_ctr_by_position_and_type = visible.groupby(['position_tier', 'content_type'])['ctr'].mean().unstack()

# Find the content_type with the minimum CTR for each position_tier
min_ctr_content_type = worst_ctr_by_position_and_type.idxmin(axis=1)
min_ctr_values = worst_ctr_by_position_and_type.min(axis=1)

# Combine into a DataFrame for display
result_df = pd.DataFrame({
    'Min CTR': min_ctr_values.round(4),
    'Worst Content Type': min_ctr_content_type
})

print("Content type with the worst CTR for each position tier (impressions >= 100):")
display(result_df)
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

```python
# Your discoveries here

## Discovery A: 'High search volume means more traffic.' Does it?

Initially, we attempted to correlate `search_volume` with pages having `impressions_90d > 0`. This resulted in a `NaN` correlation because *all* 30,000 pages in the dataset had `impressions_90d > 0`, meaning the boolean series (`df['impressions_90d'] > 0`) had no variability (all `True`). A correlation with a series of zero standard deviation is undefined.

After fixing the code to correlate `search_volume` directly with `impressions_90d` (rather than the boolean `impressions_90d > 0`), the correlation coefficient is **0.001**. This value, being very close to zero, suggests that `search_volume` barely predicts the traffic a page actually gets within this dataset. This finding challenges the belief that high search volume directly translates to high traffic.

## Discovery B: CTR cliff by position and worst content types

We analyzed which `content_type` had the worst Click-Through Rate (CTR) for each `position_tier`, considering only pages with at least 100 impressions. The findings indicate:

*   **deep:** 'feedly article' (Min CTR: 0.0440)
*   **page_1:** 'comparison article' (Min CTR: 0.1412)
*   **page_3_5:** 'comparison article' (Min CTR: 0.0940)
*   **striking:** 'comparison article' (Min CTR: 0.1474)
*   **top_3:** 'comparison article' (Min CTR: 0.0000)

'Comparison article' generally shows the worst CTR across multiple prominent position tiers, while 'feedly article' performs poorly in the 'deep' tier. This highlights that certain content types may struggle to attract clicks even when visible.

## Discovery D: Does `content_age_days` relate to `trend_direction`?

We found the median `content_age_days` for each `trend_direction`:

*   **down:** 216 days
*   **flat:** 231 days
*   **new:** 279 days
*   **stable:** 300 days
*   **up:** 292 days

It appears that pages with a 'down' trend tend to be younger on average, while 'stable' and 'up' trending pages are generally older. 'New' pages have a higher median age than 'down' or 'flat', which is an interesting observation.

## Discovery E: Does `avg_position` relate to `ctr`?

The correlation between `avg_position` and `ctr` is **-0.073**. This indicates a weak negative relationship: as average position improves (becomes a smaller number), CTR tends to slightly increase. The scatter plot visually confirms this weak trend, showing a broad distribution of CTRs across various average positions, but with a slight leaning towards higher CTRs at better (lower) average positions.
```

## Discovery D: Does `content_age_days` relate to `trend_direction`?

```python
age_by_trend = df.groupby("trend_direction")["content_age_days"].median()
print(age_by_trend.round(0).to_string())
print("\nThis shows the median content age for each trend direction. We can see if older or newer content tends to trend in a particular way.")
```

## Discovery E: Does `avg_position` relate to `ctr`?

```python
corr_pos_ctr = df["avg_position"].corr(df["ctr"])
print(f"Correlation between avg_position and ctr: {corr_pos_ctr:.3f}")
print("A negative correlation would indicate that as average position improves (becomes a smaller number), CTR tends to increase.")

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x='avg_position', y='ctr', data=df, alpha=0.5)
plt.title('Average Position vs. CTR')
plt.xlabel('Average Position')
plt.ylabel('CTR')
plt.grid(True)
plt.show()
```
